Event Eligibility Analysis

```python
# ==========================================
# MODULE 1 & MODULE 2 ELIGIBILITY AUDIT
# ==========================================

# Count CDMs in each temporal category
early = df[df["time_to_tca"] >= 2]

intermediate = df[
    (df["time_to_tca"] >= 1) &
    (df["time_to_tca"] < 2)
]

target_window = df[df["time_to_tca"] < 1]

print("=" * 60)
print("TEMPORAL DISTRIBUTION")
print("=" * 60)

print("Early CDMs (>= 2 days):", len(early))
print("Intermediate CDMs (1 to <2 days):", len(intermediate))
print("Target-window CDMs (< 1 day):", len(target_window))

# Count observations per event
early_counts = early.groupby("event_id").size()
intermediate_counts = intermediate.groupby("event_id").size()
target_counts = target_window.groupby("event_id").size()

# Create event-level audit table
event_audit = pd.DataFrame(index=df["event_id"].unique())

event_audit["early_count"] = early_counts
event_audit["intermediate_count"] = intermediate_counts
event_audit["target_count"] = target_counts

event_audit = event_audit.fillna(0).astype(int)

# Eligibility conditions
event_audit["has_minimum_early"] = (
    event_audit["early_count"] >= 3
)

event_audit["has_target"] = (
    event_audit["target_count"] >= 1
)

event_audit["module1_eligible_candidate"] = (
    event_audit["has_minimum_early"] &
    event_audit["has_target"]
)

event_audit["module2_eligible_candidate"] = (
    event_audit["has_minimum_early"] &
    event_audit["has_target"]
)

print("\n" + "=" * 60)
print("EVENT ELIGIBILITY RESULTS")
print("=" * 60)

print("Total Events:", len(event_audit))

print(
    "Events with at least 3 early CDMs:",
    event_audit["has_minimum_early"].sum()
)

print(
    "Events with at least 1 target-window CDM:",
    event_audit["has_target"].sum()
)

print(
    "Module 1 Eligible Candidates:",
    event_audit["module1_eligible_candidate"].sum()
)

print(
    "Module 2 Eligible Candidates:",
    event_audit["module2_eligible_candidate"].sum()
)

print("\nCDM COUNT SUMMARY:")
print(event_audit[
    ["early_count", "intermediate_count", "target_count"]
].describe())

print("\n" + "=" * 60)
print("ELIGIBILITY AUDIT COMPLETED")
print("=" * 60)
```

Temporal Ordering Audit

```python
# ==========================================
# CELL 15: TEMPORAL ORDERING VERIFICATION
# ==========================================

print("=" * 65)
print("TEMPORAL ORDERING AUDIT")
print("=" * 65)

# Preserve original row order for verification
df_order = df.copy()

# Original last CDM of each event
last_cdm = df_order.groupby("event_id", sort=False).tail(1)

# CDM with minimum time_to_tca for each event
min_tca_idx = df_order.groupby("event_id")["time_to_tca"].idxmin()

final_tca_cdm = df_order.loc[min_tca_idx]

# Compare original last row with minimum time_to_tca row
comparison = last_cdm[
    ["event_id", "time_to_tca", "risk"]
].merge(
    final_tca_cdm[
        ["event_id", "time_to_tca", "risk"]
    ],
    on="event_id",
    suffixes=("_last_row", "_minimum_tca")
)

# Check consistency
same_tca = np.isclose(
    comparison["time_to_tca_last_row"],
    comparison["time_to_tca_minimum_tca"]
)

print("\n1. EVENT ORDERING")
print("Total Events:", len(comparison))

print(
    "Events where last row has minimum time_to_tca:",
    same_tca.sum()
)

print(
    "Events with different temporal ordering:",
    (~same_tca).sum()
)

# Check final CDM target window
print("\n2. FINAL CDM TARGET WINDOW")

print(
    "Final CDMs with time_to_tca < 1:",
    (last_cdm["time_to_tca"] < 1).sum()
)

print(
    "Final CDMs with time_to_tca >= 1:",
    (last_cdm["time_to_tca"] >= 1).sum()
)

# Display examples
print("\n3. SAMPLE FINAL CDMs")

display(
    last_cdm[
        ["event_id", "time_to_tca", "risk"]
    ].head(10)
)

print("\n" + "=" * 65)
print("TEMPORAL ORDERING AUDIT COMPLETED")
print("=" * 65)
```

Final Eligibility Audit

```python
# ==========================================
# CELL 16: FINAL EVENT ELIGIBILITY AUDIT
# ==========================================

print("=" * 65)
print("FINAL EVENT ELIGIBILITY AUDIT")
print("=" * 65)

# 1. Get the final CDM of every event
final_cdm = df.groupby("event_id", sort=False).tail(1)

# 2. Count early CDMs for every event
early_counts = (
    df[df["time_to_tca"] >= 2]
    .groupby("event_id")
    .size()
)

# 3. Build event-level eligibility table
event_audit = pd.DataFrame({
    "event_id": df["event_id"].unique()
})

event_audit["early_cdm_count"] = (
    event_audit["event_id"]
    .map(early_counts)
    .fillna(0)
    .astype(int)
)

# 4. Add final CDM information
event_audit = event_audit.merge(
    final_cdm[["event_id", "time_to_tca", "risk"]],
    on="event_id",
    how="left"
)

event_audit = event_audit.rename(columns={
    "time_to_tca": "final_time_to_tca",
    "risk": "final_risk"
})

# 5. Apply eligibility conditions
event_audit["has_minimum_early_cdms"] = (
    event_audit["early_cdm_count"] >= 3
)

event_audit["has_valid_final_target"] = (
    event_audit["final_time_to_tca"] < 1
)

event_audit["eligible"] = (
    event_audit["has_minimum_early_cdms"] &
    event_audit["has_valid_final_target"]
)

# 6. Display results
print("\nTotal Events:", len(event_audit))

print(
    "Events with at least 3 early CDMs:",
    event_audit["has_minimum_early_cdms"].sum()
)

print(
    "Events with valid final target:",
    event_audit["has_valid_final_target"].sum()
)

print(
    "Events satisfying BOTH conditions:",
    event_audit["eligible"].sum()
)

print(
    "Events excluded:",
    (~event_audit["eligible"]).sum()
)

print("\nExclusion Breakdown:")

print(
    "Less than 3 early CDMs:",
    (~event_audit["has_minimum_early_cdms"]).sum()
)

print(
    "Final CDM outside target window:",
    (~event_audit["has_valid_final_target"]).sum()
)

print("\nSample Eligible Events:")

display(
    event_audit[event_audit["eligible"]].head(10)
)

print("\n" + "=" * 65)
print("FINAL ELIGIBILITY AUDIT COMPLETED")
print("=" * 65)
```

```python
# ==========================================
# CELL 18: SELECTED DATA MISSING VALUE AUDIT
# ==========================================

print("=" * 65)
print("SELECTED DATA MISSING VALUE AUDIT")
print("=" * 65)

# Exclude identifiers and temporal/target columns
excluded_columns = [
    "event_id",
    "time_to_tca",
    "risk"
]

feature_columns = [
    col for col in df.columns
    if col not in excluded_columns
]

# Function to inspect missing values
def missing_value_report(data, module_name):

    missing = data[feature_columns].isnull().sum()

    missing = missing[missing > 0].sort_values(
        ascending=False
    )

    total_cells = (
        len(data) * len(feature_columns)
    )

    total_missing = data[feature_columns].isnull().sum().sum()

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    print("Total Rows:", len(data))
    print("Total Features:", len(feature_columns))
    print("Total Missing Values:", total_missing)

    if total_cells > 0:
        print(
            "Missing Percentage:",
            round(total_missing / total_cells * 100, 2),
            "%"
        )

    print("\nFeatures with Missing Values:")

    if len(missing) > 0:
        report = pd.DataFrame({
            "Missing_Count": missing,
            "Missing_Percentage": (
                missing / len(data) * 100
            ).round(2)
        })

        display(report)

    else:
        print("No missing feature values found.")


# Module 1 input
missing_value_report(
    module1_input,
    "MODULE 1 INPUT"
)

# Module 2 input
missing_value_report(
    module2_input,
    "MODULE 2 INPUT"
)

# Final target
print("\n" + "-" * 60)
print("FINAL TARGET")
print("-" * 60)

print(
    "Missing Target Risk:",
    eligible_final["risk"].isna().sum()
)

print(
    "Missing Target Time:",
    eligible_final["time_to_tca"].isna().sum()
)

print("\n" + "=" * 65)
print("MISSING VALUE AUDIT COMPLETED")
print("=" * 65)
```

Missingness Pattern Analysis

```python
# ==========================================
# CELL 19: MISSINGNESS PATTERN ANALYSIS
# ==========================================

print("=" * 65)
print("MISSINGNESS PATTERN ANALYSIS")
print("=" * 65)

def analyze_missing_patterns(data, module_name):

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    # 1. Missing values per row
    row_missing = data[feature_columns].isnull().sum(axis=1)

    print("\n1. ROW-LEVEL MISSINGNESS")

    print("Rows with no missing features:", (row_missing == 0).sum())

    print("Rows with at least one missing feature:", (row_missing > 0).sum())

    print("Maximum missing features in one row:", row_missing.max())

    print(
        "Average missing features per row:",
        round(row_missing.mean(), 3)
    )

    # 2. Events affected by missing values
    affected_rows = data.loc[
        row_missing > 0,
        "event_id"
    ]

    print("\n2. EVENT-LEVEL MISSINGNESS")

    print(
        "Events with at least one missing value:",
        affected_rows.nunique()
    )

    print(
        "Total events:",
        data["event_id"].nunique()
    )

    # 3. Check important feature missingness
    important_features = [
        "c_rcs_estimate",
        "t_rcs_estimate",
        "F10",
        "F3M",
        "SSN",
        "AP"
    ]

    print("\n3. IMPORTANT FEATURE MISSINGNESS")

    for col in important_features:

        missing_rows = data[data[col].isnull()]

        print(
            f"{col}:",
            len(missing_rows),
            "rows |",
            missing_rows["event_id"].nunique(),
            "events"
        )

    # 4. Covariance feature missingness
    covariance_features = [
        col for col in feature_columns
        if "sigma" in col or "cndot" in col
        or "crdot" in col or "ctdot" in col
    ]

    covariance_missing = (
        data[covariance_features].isnull().any(axis=1)
    )

    print("\n4. COVARIANCE FEATURE MISSINGNESS")

    print(
        "Rows with at least one missing covariance feature:",
        covariance_missing.sum()
    )

    print(
        "Affected events:",
        data.loc[
            covariance_missing,
            "event_id"
        ].nunique()
    )


# Analyze both modules
analyze_missing_patterns(
    module1_input,
    "MODULE 1"
)

analyze_missing_patterns(
    module2_input,
    "MODULE 2"
)

print("\n" + "=" * 65)
print("MISSINGNESS PATTERN ANALYSIS COMPLETED")
print("=" * 65)
```

Temporal Missingness Analysis

```python
# ==========================================
# CELL 20: TEMPORAL MISSINGNESS ANALYSIS
# ==========================================

print("=" * 65)
print("TEMPORAL MISSINGNESS ANALYSIS")
print("=" * 65)

def temporal_missingness(data, module_name):

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    # Divide records according to time_to_tca
    data = data.copy()

    data["temporal_group"] = pd.cut(
        data["time_to_tca"],
        bins=[-float("inf"), 1, 2, float("inf")],
        labels=[
            "Target Window (<1 day)",
            "Intermediate (1-2 days)",
            "Early (>=2 days)"
        ],
        right=False
    )

    important_features = [
        "c_rcs_estimate",
        "t_rcs_estimate",
        "F10",
        "F3M",
        "SSN",
        "AP"
    ]

    print("\n1. MISSINGNESS BY TEMPORAL GROUP")

    for group in data["temporal_group"].dropna().unique():

        group_data = data[
            data["temporal_group"] == group
        ]

        print(f"\n{group}")
        print("Total records:", len(group_data))

        for col in important_features:

            missing = group_data[col].isnull().sum()

            percentage = (
                missing / len(group_data) * 100
                if len(group_data) > 0 else 0
            )

            print(
                f"{col}: {missing} missing "
                f"({percentage:.2f}%)"
            )

    # Check whether solar features are missing together
    print("\n2. SOLAR FEATURES CO-MISSINGNESS")

    solar_features = ["F10", "F3M", "SSN", "AP"]

    solar_all_missing = (
        data[solar_features].isnull().all(axis=1)
    )

    solar_any_missing = (
        data[solar_features].isnull().any(axis=1)
    )

    print(
        "Rows where all four solar features are missing:",
        solar_all_missing.sum()
    )

    print(
        "Rows where at least one solar feature is missing:",
        solar_any_missing.sum()
    )

    # Check c_rcs_estimate missingness at event level
    print("\n3. C_RCS EVENT-LEVEL PATTERN")

    event_c_missing = (
        data.groupby("event_id")["c_rcs_estimate"]
        .agg(
            total_records="size",
            missing_records=lambda x: x.isnull().sum()
        )
    )

    event_c_missing["missing_percentage"] = (
        event_c_missing["missing_records"]
        / event_c_missing["total_records"] * 100
    )

    print(
        "Events with all c_rcs_estimate values missing:",
        (
            event_c_missing["missing_records"]
            == event_c_missing["total_records"]
        ).sum()
    )

    print(
        "Events with partially missing c_rcs_estimate:",
        (
            (event_c_missing["missing_records"] > 0)
            &
            (
                event_c_missing["missing_records"]
                < event_c_missing["total_records"]
            )
        ).sum()
    )


# Analyze both modules

temporal_missingness(
    module1_input,
    "MODULE 1"
)

temporal_missingness(
    module2_input,
    "MODULE 2"
)

print("\n" + "=" * 65)
print("TEMPORAL MISSINGNESS ANALYSIS COMPLETED")
print("=" * 65)
```

```python
# ==========================================
# CELL 23: COMPLETE FEATURE MISSINGNESS AUDIT
# ==========================================

print("=" * 70)
print("COMPLETE FEATURE MISSINGNESS AUDIT")
print("=" * 70)


def complete_missingness_audit(data, module_name):

    print("\n" + "-" * 65)
    print(module_name)
    print("-" * 65)

    # Calculate missingness for every input feature
    missing_count = data[feature_columns].isnull().sum()

    missing_percentage = (
        missing_count / len(data) * 100
    )

    audit = pd.DataFrame({
        "missing_count": missing_count,
        "missing_percentage": missing_percentage,
        "available_count": len(data) - missing_count
    })

    # Classify features according to missingness
    def classify_missingness(percentage):

        if percentage == 0:
            return "No Missing"

        elif percentage <= 5:
            return "Low Missing (<=5%)"

        elif percentage <= 20:
            return "Moderate Missing (5-20%)"

        elif percentage <= 50:
            return "High Missing (20-50%)"

        else:
            return "Very High Missing (>50%)"

    audit["missingness_category"] = (
        audit["missing_percentage"]
        .apply(classify_missingness)
    )

    audit = audit.sort_values(
        "missing_percentage",
        ascending=False
    )

    print("\n1. MISSINGNESS CATEGORY SUMMARY")

    print(
        audit["missingness_category"]
        .value_counts()
    )

    print("\n2. COMPLETE FEATURE AUDIT")

    print(
        audit.round(2).to_string()
    )

    print("\n3. FEATURES WITH VERY HIGH MISSINGNESS")

    print(
        audit[
            audit["missing_percentage"] > 50
        ]
    )

    print("\n4. FEATURES WITH NO MISSING VALUES")

    print(
        audit[
            audit["missing_percentage"] == 0
        ].index.tolist()
    )

    return audit


# Run for both modules

module1_missingness_audit = complete_missingness_audit(
    module1_input,
    "MODULE 1"
)

module2_missingness_audit = complete_missingness_audit(
    module2_input,
    "MODULE 2"
)


print("\n" + "=" * 70)
print("COMPLETE FEATURE MISSINGNESS AUDIT FINISHED")
print("=" * 70)
```

```python
# ==========================================
# CELL 30: TEMPORAL RISK EVOLUTION AUDIT
# ==========================================

print("=" * 75)
print("TEMPORAL RISK EVOLUTION AUDIT")
print("=" * 75)

# Use eligible events only
eligible_ids = event_audit.loc[
    event_audit["eligible"],
    "event_id"
]

risk_data = df[
    df["event_id"].isin(eligible_ids)
].copy()

# Module 1 and Module 2 input observations
module1_risk = risk_data[
    risk_data["time_to_tca"] >= 2
].copy()

module2_risk = risk_data[
    risk_data["time_to_tca"] >= 1
].copy()

# Final risk of each event
final_risk = (
    risk_data.groupby("event_id", sort=False)
    .tail(1)[["event_id", "risk"]]
    .rename(columns={"risk": "final_risk"})
)

# Earliest available risk for each module
module1_earliest = (
    module1_risk.sort_values(
        ["event_id", "time_to_tca"],
        ascending=[True, False]
    )
    .groupby("event_id", sort=False)
    .tail(1)[["event_id", "risk"]]
    .rename(columns={"risk": "earliest_risk"})
)

module2_earliest = (
    module2_risk.sort_values(
        ["event_id", "time_to_tca"],
        ascending=[True, False]
    )
    .groupby("event_id", sort=False)
    .tail(1)[["event_id", "risk"]]
    .rename(columns={"risk": "earliest_risk"})
)

# Compare earliest and final risk
module1_comparison = module1_earliest.merge(
    final_risk,
    on="event_id"
)

module2_comparison = module2_earliest.merge(
    final_risk,
    on="event_id"
)

module1_comparison["risk_difference"] = (
    module1_comparison["final_risk"]
    - module1_comparison["earliest_risk"]
)

module2_comparison["risk_difference"] = (
    module2_comparison["final_risk"]
    - module2_comparison["earliest_risk"]
)

# Summary function
def risk_summary(data, module_name):

    print(f"\n{module_name}")
    print("-" * 75)

    print("Events:", len(data))

    print(
        "Mean earliest risk:",
        data["earliest_risk"].mean()
    )

    print(
        "Mean final risk:",
        data["final_risk"].mean()
    )

    print(
        "Mean risk difference:",
        data["risk_difference"].mean()
    )

    print(
        "Median absolute risk difference:",
        data["risk_difference"].abs().median()
    )

    print(
        "Risk increased:",
        (data["risk_difference"] > 0).sum()
    )

    print(
        "Risk decreased:",
        (data["risk_difference"] < 0).sum()
    )

    print(
        "Risk unchanged:",
        (data["risk_difference"] == 0).sum()
    )

    print(
        "Earliest risk = -30:",
        (data["earliest_risk"] == -30).sum()
    )

    print(
        "Final risk = -30:",
        (data["final_risk"] == -30).sum()
    )


# Print summaries
risk_summary(
    module1_comparison,
    "MODULE 1: EARLIEST CDM >= 2 DAYS"
)

risk_summary(
    module2_comparison,
    "MODULE 2: EARLIEST CDM >= 1 DAY"
)

# Overall risk distribution
print("\nOVERALL INPUT RISK DISTRIBUTION")
print("-" * 75)

print(
    "Module 1 risk = -30:",
    (module1_risk["risk"] == -30).sum()
)

print(
    "Module 2 risk = -30:",
    (module2_risk["risk"] == -30).sum()
)

print("\n" + "=" * 75)
print("TEMPORAL RISK EVOLUTION AUDIT COMPLETED")
print("=" * 75)
```

```python
# ==========================================
# CELL 31: RISK = -30 TEMPORAL PATTERN AUDIT
# ==========================================

print("=" * 75)
print("RISK = -30 TEMPORAL PATTERN INVESTIGATION")
print("=" * 75)

# Eligible events only
eligible_ids = event_audit.loc[
    event_audit["eligible"],
    "event_id"
]

risk_pattern_data = df[
    df["event_id"].isin(eligible_ids)
].copy()

# Preserve original CDM sequence
risk_pattern_data = risk_pattern_data.sort_index()

# Analyze each event
event_patterns = []

for event_id, group in risk_pattern_data.groupby(
    "event_id",
    sort=False
):

    risks = group["risk"].tolist()

    total_cdms = len(risks)

    first_risk = risks[0]

    final_risk = risks[-1]

    minus_30_positions = [
        i for i, value in enumerate(risks)
        if value == -30
    ]

    has_minus_30 = len(minus_30_positions) > 0

    first_minus_30_position = (
        minus_30_positions[0]
        if has_minus_30 else None
    )

    # Check whether -30 appears in the first CDM
    starts_with_minus_30 = first_risk == -30

    # Check whether -30 appears in the final CDM
    ends_with_minus_30 = final_risk == -30

    # Check whether risk changes after first reaching -30
    changes_after_minus_30 = False

    if has_minus_30:

        first_position = first_minus_30_position

        later_risks = risks[first_position:]

        changes_after_minus_30 = any(
            value != -30
            for value in later_risks
        )

    event_patterns.append({

        "event_id": event_id,

        "total_cdms": total_cdms,

        "first_risk": first_risk,

        "final_risk": final_risk,

        "has_minus_30": has_minus_30,

        "starts_with_minus_30": starts_with_minus_30,

        "ends_with_minus_30": ends_with_minus_30,

        "first_minus_30_position": first_minus_30_position,

        "changes_after_minus_30": changes_after_minus_30

    })


pattern_df = pd.DataFrame(event_patterns)


# ==========================================
# SUMMARY
# ==========================================

print("\n1. GENERAL RISK PATTERN")
print("-" * 75)

print("Total eligible events:", len(pattern_df))

print(
    "Events containing -30:",
    pattern_df["has_minus_30"].sum()
)

print(
    "Events starting with -30:",
    pattern_df["starts_with_minus_30"].sum()
)

print(
    "Events ending with -30:",
    pattern_df["ends_with_minus_30"].sum()
)


print("\n2. EVENTS THAT FIRST REACH -30")
print("-" * 75)

print(
    "Events reaching -30 at any CDM:",
    pattern_df["has_minus_30"].sum()
)

print(
    "Events changing after reaching -30:",
    pattern_df["changes_after_minus_30"].sum()
)

print(
    "Events maintaining -30 after first occurrence:",
    (
        pattern_df["has_minus_30"]
        & ~pattern_df["changes_after_minus_30"]
    ).sum()
)


print("\n3. FINAL -30 ANALYSIS")
print("-" * 75)

final_minus_30 = pattern_df[
    pattern_df["final_risk"] == -30
]

print(
    "Events with final risk -30:",
    len(final_minus_30)
)

print(
    "Among them, events starting with -30:",
    final_minus_30["starts_with_minus_30"].sum()
)

print(
    "Among them, events reaching -30 later:",
    (
        ~final_minus_30["starts_with_minus_30"]
    ).sum()
)


print("\n4. FIRST OCCURRENCE POSITION")
print("-" * 75)

print(
    pattern_df.loc[
        pattern_df["has_minus_30"],
        "first_minus_30_position"
    ].describe()
)


print("\n5. SAMPLE EVENTS WITH FINAL RISK -30")
print("-" * 75)

print(
    pattern_df[
        pattern_df["final_risk"] == -30
    ].head(10).to_string(index=False)
)


print("\n" + "=" * 75)
print("RISK = -30 TEMPORAL PATTERN AUDIT COMPLETED")
print("=" * 75)
```

```python
# ==========================================
# CELL 32: RISK TRANSITION ANALYSIS
# ==========================================

print("=" * 75)
print("RISK TRANSITION ANALYSIS")
print("=" * 75)

# Eligible events only
eligible_ids = event_audit.loc[
    event_audit["eligible"],
    "event_id"
]

transition_data = df[
    df["event_id"].isin(eligible_ids)
].copy()

# Preserve original CDM sequence
transition_data = transition_data.sort_index()

event_transition_results = []

total_transitions = 0
transitions_to_minus_30 = 0
transitions_from_minus_30 = 0
transitions_between_other_values = 0

for event_id, group in transition_data.groupby(
    "event_id",
    sort=False
):

    risks = group["risk"].tolist()

    changes = 0
    to_minus_30 = 0
    from_minus_30 = 0
    other_changes = 0

    for i in range(1, len(risks)):

        previous_risk = risks[i - 1]
        current_risk = risks[i]

        if previous_risk != current_risk:

            changes += 1

            if previous_risk != -30 and current_risk == -30:

                to_minus_30 += 1

            elif previous_risk == -30 and current_risk != -30:

                from_minus_30 += 1

            else:

                other_changes += 1

    total_transitions += changes
    transitions_to_minus_30 += to_minus_30
    transitions_from_minus_30 += from_minus_30
    transitions_between_other_values += other_changes

    event_transition_results.append({

        "event_id": event_id,

        "total_cdms": len(risks),

        "risk_changes": changes,

        "transitions_to_minus_30": to_minus_30,

        "transitions_from_minus_30": from_minus_30,

        "other_risk_changes": other_changes,

        "unique_risk_values": len(set(risks)),

        "is_monotonic": (
            all(risks[i] <= risks[i - 1]
                for i in range(1, len(risks)))
            or
            all(risks[i] >= risks[i - 1]
                for i in range(1, len(risks)))
        )

    })


transition_df = pd.DataFrame(event_transition_results)


# ==========================================
# SUMMARY
# ==========================================

print("\n1. OVERALL TRANSITIONS")
print("-" * 75)

print("Total events:", len(transition_df))

print("Total consecutive risk changes:", total_transitions)

print("Transitions TO -30:", transitions_to_minus_30)

print("Transitions FROM -30:", transitions_from_minus_30)

print(
    "Changes between non -30 values:",
    transitions_between_other_values
)


print("\n2. EVENT-LEVEL BEHAVIOUR")
print("-" * 75)

print(
    "Events with no risk changes:",
    (transition_df["risk_changes"] == 0).sum()
)

print(
    "Events with at least one risk change:",
    (transition_df["risk_changes"] > 0).sum()
)

print(
    "Events with multiple risk changes:",
    (transition_df["risk_changes"] > 1).sum()
)

print(
    "Events with non-monotonic risk sequence:",
    (~transition_df["is_monotonic"]).sum()
)


print("\n3. RISK CHANGE STATISTICS")
print("-" * 75)

print(
    transition_df["risk_changes"].describe()
)


print("\n4. SAMPLE EVENTS WITH MULTIPLE RISK CHANGES")
print("-" * 75)

print(
    transition_df[
        transition_df["risk_changes"] > 1
    ].head(10).to_string(index=False)
)


print("\n" + "=" * 75)
print("RISK TRANSITION ANALYSIS COMPLETED")
print("=" * 75)
```

Numerical Feature Validity & Outlier Audit

```python
# ==========================================
# CELL 33: NUMERICAL FEATURE VALIDITY AUDIT
# ==========================================

print("=" * 75)
print("NUMERICAL FEATURE VALIDITY AND OUTLIER AUDIT")
print("=" * 75)

import numpy as np

# Features requiring special numerical investigation
audit_columns = [
    "miss_distance",
    "relative_speed",
    "mahalanobis_distance",
    "t_position_covariance_det",
    "c_position_covariance_det",
    "t_sigma_r",
    "c_sigma_r",
    "t_weighted_rms",
    "c_weighted_rms",
    "max_risk_estimate",
    "max_risk_scaling"
]

audit_columns = [
    col for col in audit_columns if col in df.columns
]

results = []

for col in audit_columns:

    series = pd.to_numeric(df[col], errors="coerce")

    finite_values = series[np.isfinite(series)]

    q1 = finite_values.quantile(0.25)
    q3 = finite_values.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    results.append({

        "feature": col,

        "missing": int(series.isna().sum()),

        "infinite": int(np.isinf(series).sum()),

        "negative": int((finite_values < 0).sum()),

        "zero": int((finite_values == 0).sum()),

        "minimum": finite_values.min(),

        "maximum": finite_values.max(),

        "outlier_count": int(
            (
                (finite_values < lower_bound)
                | (finite_values > upper_bound)
            ).sum()
        )

    })


validity_df = pd.DataFrame(results)

print("\n1. NUMERICAL VALIDITY SUMMARY")
print("-" * 75)

print(validity_df.to_string(index=False))


print("\n2. COVARIANCE DETERMINANT INVESTIGATION")
print("-" * 75)

for col in [
    "t_position_covariance_det",
    "c_position_covariance_det"
]:

    if col in df.columns:

        negative_values = df.loc[
            df[col] < 0,
            ["event_id", "time_to_tca", col]
        ]

        print(f"\n{col}")

        print("Negative records:", len(negative_values))

        print("Affected events:", negative_values["event_id"].nunique())

        print("Sample negative values:")

        print(negative_values.head(5).to_string(index=False))


print("\n3. EXTREME VALUE PERCENTILES")
print("-" * 75)

for col in audit_columns:

    print(f"\n{col}")

    print(
        df[col].quantile(
            [0, 0.01, 0.25, 0.50, 0.75, 0.99, 1]
        ).to_string()
    )


print("\n" + "=" * 75)
print("NUMERICAL FEATURE VALIDITY AUDIT COMPLETED")
print("=" * 75)
```

```python
# ==========================================
# CELL 37: FINAL DATASET FILTRATION
# ==========================================

print("=" * 75)
print("FINAL DATASET FILTRATION")
print("=" * 75)


# ------------------------------------------
# 1. IDENTIFY ELIGIBLE EVENTS
# ------------------------------------------

# Original last CDM of each event
final_rows = (
    df.groupby("event_id", sort=False)
    .tail(1)
    .copy()
)

# Count early CDMs for every event
early_counts = (
    df[df["time_to_tca"] >= 2]
    .groupby("event_id")
    .size()
)

# Eligible events:
# Minimum 3 early CDMs
# Final CDM must be within 1 day of TCA

final_rows["early_count"] = (
    final_rows["event_id"]
    .map(early_counts)
    .fillna(0)
)

eligible_targets = final_rows[
    (final_rows["early_count"] >= 3)
    & (final_rows["time_to_tca"] < 1)
].copy()

eligible_ids = set(eligible_targets["event_id"])

print("\n1. ELIGIBLE EVENTS")
print("-" * 75)

print("Total original events:", df["event_id"].nunique())
print("Eligible events:", len(eligible_ids))
print("Excluded events:", df["event_id"].nunique() - len(eligible_ids))


# ------------------------------------------
# 2. CREATE FINAL TARGET DATASET
# ------------------------------------------

targets = eligible_targets[
    ["event_id", "time_to_tca", "risk"]
].copy()

targets = targets.rename(columns={
    "time_to_tca": "final_time_to_tca",
    "risk": "final_risk"
})

targets = targets.sort_values("event_id").reset_index(drop=True)


# ------------------------------------------
# 3. CREATE MODULE 1 INPUT
# ------------------------------------------

module1_input = df[
    (df["event_id"].isin(eligible_ids))
    & (df["time_to_tca"] >= 2)
].copy()

# Earliest CDM first, latest input CDM last
module1_input = module1_input.sort_values(
    ["event_id", "time_to_tca"],
    ascending=[True, False]
).reset_index(drop=True)

module1_input = module1_input.rename(columns={
    "risk": "current_risk"
})


# ------------------------------------------
# 4. CREATE MODULE 2 INPUT
# ------------------------------------------

module2_input = df[
    (df["event_id"].isin(eligible_ids))
    & (df["time_to_tca"] >= 1)
].copy()

# Earliest CDM first, latest input CDM last
module2_input = module2_input.sort_values(
    ["event_id", "time_to_tca"],
    ascending=[True, False]
).reset_index(drop=True)

module2_input = module2_input.rename(columns={
    "risk": "current_risk"
})


# ------------------------------------------
# 5. VALIDATION
# ------------------------------------------

print("\n2. MODULE DATASET SUMMARY")
print("-" * 75)

print("\nModule 1:")
print("Input records:", len(module1_input))
print("Unique events:", module1_input["event_id"].nunique())

print("\nModule 2:")
print("Input records:", len(module2_input))
print("Unique events:", module2_input["event_id"].nunique())

print("\nTargets:")
print("Target records:", len(targets))
print("Unique target events:", targets["event_id"].nunique())


# Check minimum input CDMs
m1_counts = module1_input.groupby("event_id").size()
m2_counts = module2_input.groupby("event_id").size()

print("\n3. VALIDATION CHECKS")
print("-" * 75)

print(
    "Module 1: All events have >= 3 inputs:",
    (m1_counts >= 3).all()
)

print(
    "Module 2: All events have >= 3 inputs:",
    (m2_counts >= 3).all()
)

print(
    "All target times < 1 day:",
    (targets["final_time_to_tca"] < 1).all()
)

print(
    "Module 1 and target event IDs match:",
    set(module1_input["event_id"]) == set(targets["event_id"])
)

print(
    "Module 2 and target event IDs match:",
    set(module2_input["event_id"]) == set(targets["event_id"])
)

print(
    "Final risk missing values:",
    targets["final_risk"].isna().sum()
)


# ------------------------------------------
# 6. SAVE FINAL DATASETS
# ------------------------------------------

OUTPUT_DIR = PROJECT_DIR / "Results" / "Filtered_Datasets"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

module1_input.to_csv(
    OUTPUT_DIR / "Module1_Early_Input.csv",
    index=False
)

module2_input.to_csv(
    OUTPUT_DIR / "Module2_Updated_Input.csv",
    index=False
)

targets.to_csv(
    OUTPUT_DIR / "Final_Risk_Targets.csv",
    index=False
)

print("\n4. SAVED FILES")
print("-" * 75)

print("Output folder:", OUTPUT_DIR)

print("1. Module1_Early_Input.csv")
print("2. Module2_Updated_Input.csv")
print("3. Final_Risk_Targets.csv")

print("\n" + "=" * 75)
print("FINAL DATASET FILTRATION COMPLETED")
print("=" * 75)
```

```python
# ==========================================
# CELL 41: EVENT-LEVEL FEATURE ENGINEERING
# ==========================================

import numpy as np
import pandas as pd


def create_event_features(data, module_name):

    print(f"\nProcessing {module_name}...")

    # Sort CDMs chronologically:
    # Largest time_to_tca = earliest CDM
    # Smallest time_to_tca = latest CDM

    data = data.sort_values(
        ["event_id", "time_to_tca"],
        ascending=[True, False]
    ).copy()

    # Identify feature types
    numeric_cols = data.select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_cols = data.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    # event_id is not a model feature
    numeric_cols = [
        col for col in numeric_cols
        if col != "event_id"
    ]

    categorical_cols = [
        col for col in categorical_cols
        if col != "event_id"
    ]

    # Numeric aggregation
    numeric_features = data.groupby("event_id")[
        numeric_cols
    ].agg([
        "first",
        "last",
        "mean",
        "std"
    ])

    numeric_features.columns = [
        f"{col}_{stat}"
        for col, stat in numeric_features.columns
    ]

    # Categorical aggregation
    if categorical_cols:

        categorical_features = data.groupby(
            "event_id"
        )[categorical_cols].agg([
            "first",
            "last"
        ])

        categorical_features.columns = [
            f"{col}_{stat}"
            for col, stat in categorical_features.columns
        ]

    else:
        categorical_features = pd.DataFrame(
            index=numeric_features.index
        )

    # Number of CDMs per event
    sequence_length = data.groupby(
        "event_id"
    ).size().rename("sequence_length")

    # Combine all features
    event_features = pd.concat(
        [
            numeric_features,
            categorical_features,
            sequence_length
        ],
        axis=1
    )

    event_features = event_features.reset_index()

    print("Original CDM Records:", len(data))
    print("Unique Events:", event_features["event_id"].nunique())
    print("Engineered Features:", event_features.shape[1] - 1)

    return event_features


# Create Module 1 features
X_m1_train = create_event_features(
    m1_train,
    "MODULE 1 TRAIN"
)

X_m1_test = create_event_features(
    m1_test,
    "MODULE 1 TEST"
)


# Create Module 2 features
X_m2_train = create_event_features(
    m2_train,
    "MODULE 2 TRAIN"
)

X_m2_test = create_event_features(
    m2_test,
    "MODULE 2 TEST"
)


print("\n" + "=" * 70)
print("FEATURE ENGINEERING COMPLETED")
print("=" * 70)

print("\nMODULE 1")
print("Train:", X_m1_train.shape)
print("Test:", X_m1_test.shape)

print("\nMODULE 2")
print("Train:", X_m2_train.shape)
print("Test:", X_m2_test.shape)
```

```python
# ==========================================
# CELL 42: CORRECT TEMPORAL FEATURE ENGINEERING
# ==========================================

import numpy as np
import pandas as pd


def create_event_features(data, module_name):

    print(f"\nProcessing {module_name}...")

    # Chronological sorting
    data = data.sort_values(
        ["event_id", "time_to_tca"],
        ascending=[True, False]
    ).copy()

    # Identify feature columns
    feature_cols = [
        col for col in data.columns
        if col != "event_id"
    ]

    numeric_cols = data[feature_cols].select_dtypes(
        include=np.number
    ).columns.tolist()

    categorical_cols = [
        col for col in feature_cols
        if col not in numeric_cols
    ]

    # Actual earliest CDM of each event
    earliest = (
        data.groupby("event_id", sort=False)
        .head(1)
        .set_index("event_id")
    )

    # Actual latest input CDM of each event
    latest = (
        data.groupby("event_id", sort=False)
        .tail(1)
        .set_index("event_id")
    )

    # Preserve actual first and last CDM values,
    # including missing values
    first_features = earliest[feature_cols].copy()
    first_features.columns = [
        f"{col}_first" for col in feature_cols
    ]

    last_features = latest[feature_cols].copy()
    last_features.columns = [
        f"{col}_last" for col in feature_cols
    ]

    # Numeric statistical features
    numeric_stats = data.groupby("event_id")[numeric_cols].agg(
        ["mean", "std"]
    )

    numeric_stats.columns = [
        f"{col}_{stat}"
        for col, stat in numeric_stats.columns
    ]

    # Sequence length
    sequence_length = data.groupby(
        "event_id"
    ).size().rename("sequence_length")

    # Combine features
    event_features = pd.concat(
        [
            first_features,
            last_features,
            numeric_stats,
            sequence_length
        ],
        axis=1
    )

    event_features = event_features.reset_index()

    print("Original Records:", len(data))
    print("Unique Events:", event_features["event_id"].nunique())
    print("Feature Count:", event_features.shape[1] - 1)

    return event_features


# Module 1
X_m1_train = create_event_features(
    m1_train, "MODULE 1 TRAIN"
)

X_m1_test = create_event_features(
    m1_test, "MODULE 1 TEST"
)


# Module 2
X_m2_train = create_event_features(
    m2_train, "MODULE 2 TRAIN"
)

X_m2_test = create_event_features(
    m2_test, "MODULE 2 TEST"
)


print("\n" + "=" * 70)
print("CORRECTED FEATURE ENGINEERING COMPLETED")
print("=" * 70)

print("\nMODULE 1")
print("Train:", X_m1_train.shape)
print("Test:", X_m1_test.shape)

print("\nMODULE 2")
print("Train:", X_m2_train.shape)
print("Test:", X_m2_test.shape)
```

```python
# ==========================================
# CELL 43: FINAL FEATURE VERIFICATION
# ==========================================

def verify_dataset(name, X_train, X_test, y_train, y_test):

    print("\n" + "=" * 75)
    print(name)
    print("=" * 75)

    # Feature columns
    train_cols = set(X_train.columns) - {"event_id"}
    test_cols = set(X_test.columns) - {"event_id"}

    print("\n1. FEATURE COLUMNS")
    print("Training Features:", len(train_cols))
    print("Testing Features:", len(test_cols))
    print("Same Columns:", train_cols == test_cols)

    # Leakage check
    print("\n2. TARGET LEAKAGE CHECK")
    print("final_risk in features:",
          "final_risk" in train_cols)

    # Missing values
    X_train_features = X_train.drop(columns=["event_id"])
    X_test_features = X_test.drop(columns=["event_id"])

    print("\n3. MISSING VALUES")
    print("Training Missing:", X_train_features.isna().sum().sum())
    print("Testing Missing:", X_test_features.isna().sum().sum())

    # Categorical features
    categorical = X_train_features.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    print("\n4. CATEGORICAL FEATURES")
    print("Categorical Feature Count:", len(categorical))
    print("Categorical Columns:", categorical)

    # Event ID alignment
    print("\n5. EVENT ID ALIGNMENT")

    print("Training IDs Match:",
          set(X_train["event_id"]) == set(y_train["event_id"]))

    print("Testing IDs Match:",
          set(X_test["event_id"]) == set(y_test["event_id"]))

    # Target missing values
    print("\n6. TARGET CHECK")
    print("Training Target Missing:",
          y_train["final_risk"].isna().sum())

    print("Testing Target Missing:",
          y_test["final_risk"].isna().sum())


# Verify Module 1
verify_dataset(
    "MODULE 1",
    X_m1_train,
    X_m1_test,
    y_train,
    y_test
)

# Verify Module 2
verify_dataset(
    "MODULE 2",
    X_m2_train,
    X_m2_test,
    y_train,
    y_test
)

print("\n" + "=" * 75)
print("FINAL VERIFICATION COMPLETED")
print("=" * 75)
```
