Event Eligibility Analysis

```python
# ==========================================
# MODULE 1 & MODULE 2 ELIGIBILITY AUDIT
# ==========================================

# Count CDMs in each temporal category
early = df[df["time_to_tca"] >= 2]

intermediate = df[
    (df["time_to_tca"] >= 1) &
    (df["time_to_tca"] < 2)
]

target_window = df[df["time_to_tca"] < 1]

print("=" * 60)
print("TEMPORAL DISTRIBUTION")
print("=" * 60)

print("Early CDMs (>= 2 days):", len(early))
print("Intermediate CDMs (1 to <2 days):", len(intermediate))
print("Target-window CDMs (< 1 day):", len(target_window))

# Count observations per event
early_counts = early.groupby("event_id").size()
intermediate_counts = intermediate.groupby("event_id").size()
target_counts = target_window.groupby("event_id").size()

# Create event-level audit table
event_audit = pd.DataFrame(index=df["event_id"].unique())

event_audit["early_count"] = early_counts
event_audit["intermediate_count"] = intermediate_counts
event_audit["target_count"] = target_counts

event_audit = event_audit.fillna(0).astype(int)

# Eligibility conditions
event_audit["has_minimum_early"] = (
    event_audit["early_count"] >= 3
)

event_audit["has_target"] = (
    event_audit["target_count"] >= 1
)

event_audit["module1_eligible_candidate"] = (
    event_audit["has_minimum_early"] &
    event_audit["has_target"]
)

event_audit["module2_eligible_candidate"] = (
    event_audit["has_minimum_early"] &
    event_audit["has_target"]
)

print("\n" + "=" * 60)
print("EVENT ELIGIBILITY RESULTS")
print("=" * 60)

print("Total Events:", len(event_audit))

print(
    "Events with at least 3 early CDMs:",
    event_audit["has_minimum_early"].sum()
)

print(
    "Events with at least 1 target-window CDM:",
    event_audit["has_target"].sum()
)

print(
    "Module 1 Eligible Candidates:",
    event_audit["module1_eligible_candidate"].sum()
)

print(
    "Module 2 Eligible Candidates:",
    event_audit["module2_eligible_candidate"].sum()
)

print("\nCDM COUNT SUMMARY:")
print(event_audit[
    ["early_count", "intermediate_count", "target_count"]
].describe())

print("\n" + "=" * 60)
print("ELIGIBILITY AUDIT COMPLETED")
print("=" * 60)
```

Temporal Ordering Audit

```python
# ==========================================
# CELL 15: TEMPORAL ORDERING VERIFICATION
# ==========================================

print("=" * 65)
print("TEMPORAL ORDERING AUDIT")
print("=" * 65)

# Preserve original row order for verification
df_order = df.copy()

# Original last CDM of each event
last_cdm = df_order.groupby("event_id", sort=False).tail(1)

# CDM with minimum time_to_tca for each event
min_tca_idx = df_order.groupby("event_id")["time_to_tca"].idxmin()

final_tca_cdm = df_order.loc[min_tca_idx]

# Compare original last row with minimum time_to_tca row
comparison = last_cdm[
    ["event_id", "time_to_tca", "risk"]
].merge(
    final_tca_cdm[
        ["event_id", "time_to_tca", "risk"]
    ],
    on="event_id",
    suffixes=("_last_row", "_minimum_tca")
)

# Check consistency
same_tca = np.isclose(
    comparison["time_to_tca_last_row"],
    comparison["time_to_tca_minimum_tca"]
)

print("\n1. EVENT ORDERING")
print("Total Events:", len(comparison))

print(
    "Events where last row has minimum time_to_tca:",
    same_tca.sum()
)

print(
    "Events with different temporal ordering:",
    (~same_tca).sum()
)

# Check final CDM target window
print("\n2. FINAL CDM TARGET WINDOW")

print(
    "Final CDMs with time_to_tca < 1:",
    (last_cdm["time_to_tca"] < 1).sum()
)

print(
    "Final CDMs with time_to_tca >= 1:",
    (last_cdm["time_to_tca"] >= 1).sum()
)

# Display examples
print("\n3. SAMPLE FINAL CDMs")

display(
    last_cdm[
        ["event_id", "time_to_tca", "risk"]
    ].head(10)
)

print("\n" + "=" * 65)
print("TEMPORAL ORDERING AUDIT COMPLETED")
print("=" * 65)
```

Final Eligibility Audit

```python
# ==========================================
# CELL 16: FINAL EVENT ELIGIBILITY AUDIT
# ==========================================

print("=" * 65)
print("FINAL EVENT ELIGIBILITY AUDIT")
print("=" * 65)

# 1. Get the final CDM of every event
final_cdm = df.groupby("event_id", sort=False).tail(1)

# 2. Count early CDMs for every event
early_counts = (
    df[df["time_to_tca"] >= 2]
    .groupby("event_id")
    .size()
)

# 3. Build event-level eligibility table
event_audit = pd.DataFrame({
    "event_id": df["event_id"].unique()
})

event_audit["early_cdm_count"] = (
    event_audit["event_id"]
    .map(early_counts)
    .fillna(0)
    .astype(int)
)

# 4. Add final CDM information
event_audit = event_audit.merge(
    final_cdm[["event_id", "time_to_tca", "risk"]],
    on="event_id",
    how="left"
)

event_audit = event_audit.rename(columns={
    "time_to_tca": "final_time_to_tca",
    "risk": "final_risk"
})

# 5. Apply eligibility conditions
event_audit["has_minimum_early_cdms"] = (
    event_audit["early_cdm_count"] >= 3
)

event_audit["has_valid_final_target"] = (
    event_audit["final_time_to_tca"] < 1
)

event_audit["eligible"] = (
    event_audit["has_minimum_early_cdms"] &
    event_audit["has_valid_final_target"]
)

# 6. Display results
print("\nTotal Events:", len(event_audit))

print(
    "Events with at least 3 early CDMs:",
    event_audit["has_minimum_early_cdms"].sum()
)

print(
    "Events with valid final target:",
    event_audit["has_valid_final_target"].sum()
)

print(
    "Events satisfying BOTH conditions:",
    event_audit["eligible"].sum()
)

print(
    "Events excluded:",
    (~event_audit["eligible"]).sum()
)

print("\nExclusion Breakdown:")

print(
    "Less than 3 early CDMs:",
    (~event_audit["has_minimum_early_cdms"]).sum()
)

print(
    "Final CDM outside target window:",
    (~event_audit["has_valid_final_target"]).sum()
)

print("\nSample Eligible Events:")

display(
    event_audit[event_audit["eligible"]].head(10)
)

print("\n" + "=" * 65)
print("FINAL ELIGIBILITY AUDIT COMPLETED")
print("=" * 65)
```

```python
# ==========================================
# CELL 18: SELECTED DATA MISSING VALUE AUDIT
# ==========================================

print("=" * 65)
print("SELECTED DATA MISSING VALUE AUDIT")
print("=" * 65)

# Exclude identifiers and temporal/target columns
excluded_columns = [
    "event_id",
    "time_to_tca",
    "risk"
]

feature_columns = [
    col for col in df.columns
    if col not in excluded_columns
]

# Function to inspect missing values
def missing_value_report(data, module_name):

    missing = data[feature_columns].isnull().sum()

    missing = missing[missing > 0].sort_values(
        ascending=False
    )

    total_cells = (
        len(data) * len(feature_columns)
    )

    total_missing = data[feature_columns].isnull().sum().sum()

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    print("Total Rows:", len(data))
    print("Total Features:", len(feature_columns))
    print("Total Missing Values:", total_missing)

    if total_cells > 0:
        print(
            "Missing Percentage:",
            round(total_missing / total_cells * 100, 2),
            "%"
        )

    print("\nFeatures with Missing Values:")

    if len(missing) > 0:
        report = pd.DataFrame({
            "Missing_Count": missing,
            "Missing_Percentage": (
                missing / len(data) * 100
            ).round(2)
        })

        display(report)

    else:
        print("No missing feature values found.")


# Module 1 input
missing_value_report(
    module1_input,
    "MODULE 1 INPUT"
)

# Module 2 input
missing_value_report(
    module2_input,
    "MODULE 2 INPUT"
)

# Final target
print("\n" + "-" * 60)
print("FINAL TARGET")
print("-" * 60)

print(
    "Missing Target Risk:",
    eligible_final["risk"].isna().sum()
)

print(
    "Missing Target Time:",
    eligible_final["time_to_tca"].isna().sum()
)

print("\n" + "=" * 65)
print("MISSING VALUE AUDIT COMPLETED")
print("=" * 65)
```

Missingness Pattern Analysis

```python
# ==========================================
# CELL 19: MISSINGNESS PATTERN ANALYSIS
# ==========================================

print("=" * 65)
print("MISSINGNESS PATTERN ANALYSIS")
print("=" * 65)

def analyze_missing_patterns(data, module_name):

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    # 1. Missing values per row
    row_missing = data[feature_columns].isnull().sum(axis=1)

    print("\n1. ROW-LEVEL MISSINGNESS")

    print("Rows with no missing features:", (row_missing == 0).sum())

    print("Rows with at least one missing feature:", (row_missing > 0).sum())

    print("Maximum missing features in one row:", row_missing.max())

    print(
        "Average missing features per row:",
        round(row_missing.mean(), 3)
    )

    # 2. Events affected by missing values
    affected_rows = data.loc[
        row_missing > 0,
        "event_id"
    ]

    print("\n2. EVENT-LEVEL MISSINGNESS")

    print(
        "Events with at least one missing value:",
        affected_rows.nunique()
    )

    print(
        "Total events:",
        data["event_id"].nunique()
    )

    # 3. Check important feature missingness
    important_features = [
        "c_rcs_estimate",
        "t_rcs_estimate",
        "F10",
        "F3M",
        "SSN",
        "AP"
    ]

    print("\n3. IMPORTANT FEATURE MISSINGNESS")

    for col in important_features:

        missing_rows = data[data[col].isnull()]

        print(
            f"{col}:",
            len(missing_rows),
            "rows |",
            missing_rows["event_id"].nunique(),
            "events"
        )

    # 4. Covariance feature missingness
    covariance_features = [
        col for col in feature_columns
        if "sigma" in col or "cndot" in col
        or "crdot" in col or "ctdot" in col
    ]

    covariance_missing = (
        data[covariance_features].isnull().any(axis=1)
    )

    print("\n4. COVARIANCE FEATURE MISSINGNESS")

    print(
        "Rows with at least one missing covariance feature:",
        covariance_missing.sum()
    )

    print(
        "Affected events:",
        data.loc[
            covariance_missing,
            "event_id"
        ].nunique()
    )


# Analyze both modules
analyze_missing_patterns(
    module1_input,
    "MODULE 1"
)

analyze_missing_patterns(
    module2_input,
    "MODULE 2"
)

print("\n" + "=" * 65)
print("MISSINGNESS PATTERN ANALYSIS COMPLETED")
print("=" * 65)
```

Temporal Missingness Analysis

```python
# ==========================================
# CELL 20: TEMPORAL MISSINGNESS ANALYSIS
# ==========================================

print("=" * 65)
print("TEMPORAL MISSINGNESS ANALYSIS")
print("=" * 65)

def temporal_missingness(data, module_name):

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    # Divide records according to time_to_tca
    data = data.copy()

    data["temporal_group"] = pd.cut(
        data["time_to_tca"],
        bins=[-float("inf"), 1, 2, float("inf")],
        labels=[
            "Target Window (<1 day)",
            "Intermediate (1-2 days)",
            "Early (>=2 days)"
        ],
        right=False
    )

    important_features = [
        "c_rcs_estimate",
        "t_rcs_estimate",
        "F10",
        "F3M",
        "SSN",
        "AP"
    ]

    print("\n1. MISSINGNESS BY TEMPORAL GROUP")

    for group in data["temporal_group"].dropna().unique():

        group_data = data[
            data["temporal_group"] == group
        ]

        print(f"\n{group}")
        print("Total records:", len(group_data))

        for col in important_features:

            missing = group_data[col].isnull().sum()

            percentage = (
                missing / len(group_data) * 100
                if len(group_data) > 0 else 0
            )

            print(
                f"{col}: {missing} missing "
                f"({percentage:.2f}%)"
            )

    # Check whether solar features are missing together
    print("\n2. SOLAR FEATURES CO-MISSINGNESS")

    solar_features = ["F10", "F3M", "SSN", "AP"]

    solar_all_missing = (
        data[solar_features].isnull().all(axis=1)
    )

    solar_any_missing = (
        data[solar_features].isnull().any(axis=1)
    )

    print(
        "Rows where all four solar features are missing:",
        solar_all_missing.sum()
    )

    print(
        "Rows where at least one solar feature is missing:",
        solar_any_missing.sum()
    )

    # Check c_rcs_estimate missingness at event level
    print("\n3. C_RCS EVENT-LEVEL PATTERN")

    event_c_missing = (
        data.groupby("event_id")["c_rcs_estimate"]
        .agg(
            total_records="size",
            missing_records=lambda x: x.isnull().sum()
        )
    )

    event_c_missing["missing_percentage"] = (
        event_c_missing["missing_records"]
        / event_c_missing["total_records"] * 100
    )

    print(
        "Events with all c_rcs_estimate values missing:",
        (
            event_c_missing["missing_records"]
            == event_c_missing["total_records"]
        ).sum()
    )

    print(
        "Events with partially missing c_rcs_estimate:",
        (
            (event_c_missing["missing_records"] > 0)
            &
            (
                event_c_missing["missing_records"]
                < event_c_missing["total_records"]
            )
        ).sum()
    )


# Analyze both modules

temporal_missingness(
    module1_input,
    "MODULE 1"
)

temporal_missingness(
    module2_input,
    "MODULE 2"
)

print("\n" + "=" * 65)
print("TEMPORAL MISSINGNESS ANALYSIS COMPLETED")
print("=" * 65)
```

Missingness by Object Type

```python
# ==========================================
# CELL 21: MISSINGNESS BY OBJECT TYPE
# ==========================================

print("=" * 65)
print("MISSINGNESS BY OBJECT TYPE")
print("=" * 65)


def analyze_object_type(data, module_name):

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    # Check available object types
    print("\n1. AVAILABLE OBJECT TYPES")

    print(
        data["c_object_type"]
        .value_counts(dropna=False)
    )

    # Analyze missingness by object type
    print("\n2. C_RCS_ESTIMATE MISSINGNESS BY OBJECT TYPE")

    result = (
        data.groupby(
            "c_object_type",
            dropna=False
        )
        .agg(
            total_records=("c_rcs_estimate", "size"),
            missing_records=("c_rcs_estimate",
                             lambda x: x.isnull().sum()),
            available_records=("c_rcs_estimate",
                               lambda x: x.notnull().sum())
        )
    )

    result["missing_percentage"] = (
        result["missing_records"]
        / result["total_records"] * 100
    )

    print(result.round(2))

    # Analyze t_rcs_estimate missingness
    print("\n3. T_RCS_ESTIMATE MISSINGNESS BY OBJECT TYPE")

    t_result = (
        data.groupby(
            "c_object_type",
            dropna=False
        )
        .agg(
            total_records=("t_rcs_estimate", "size"),
            missing_records=("t_rcs_estimate",
                             lambda x: x.isnull().sum())
        )
    )

    t_result["missing_percentage"] = (
        t_result["missing_records"]
        / t_result["total_records"] * 100
    )

    print(t_result.round(2))


# Analyze both modules

analyze_object_type(
    module1_input,
    "MODULE 1"
)

analyze_object_type(
    module2_input,
    "MODULE 2"
)

print("\n" + "=" * 65)
print("OBJECT TYPE ANALYSIS COMPLETED")
print("=" * 65)
```

Object Type Consistency by Event

```python
# ==========================================
# CELL 22: OBJECT TYPE CONSISTENCY BY EVENT
# ==========================================

print("=" * 65)
print("OBJECT TYPE CONSISTENCY BY EVENT")
print("=" * 65)


def check_object_consistency(data, module_name):

    print("\n" + "-" * 60)
    print(module_name)
    print("-" * 60)

    # Count unique object types per event
    event_object_types = (
        data.groupby("event_id")["c_object_type"]
        .nunique(dropna=False)
    )

    consistent_events = (
        event_object_types == 1
    ).sum()

    inconsistent_events = (
        event_object_types > 1
    ).sum()

    print("\n1. OBJECT TYPE CONSISTENCY")

    print("Total events:", len(event_object_types))

    print("Events with consistent object type:", consistent_events)

    print("Events with changing object type:", inconsistent_events)

    # Show inconsistent event samples
    print("\n2. INCONSISTENT EVENT SAMPLES")

    inconsistent_ids = event_object_types[
        event_object_types > 1
    ].index

    if len(inconsistent_ids) > 0:

        samples = data[
            data["event_id"].isin(inconsistent_ids)
        ][
            ["event_id", "time_to_tca", "c_object_type"]
        ]

        print(samples.head(20).to_string(index=False))

    else:

        print("No inconsistent events found.")

    # Check UNKNOWN events
    print("\n3. UNKNOWN OBJECT TYPE EVENTS")

    unknown_records = data[
        data["c_object_type"] == "UNKNOWN"
    ]

    print(
        "Records with UNKNOWN type:",
        len(unknown_records)
    )

    print(
        "Events containing UNKNOWN type:",
        unknown_records["event_id"].nunique()
    )

    print(
        "UNKNOWN events with available c_rcs_estimate:",
        unknown_records[
            "c_rcs_estimate"
        ].notnull().groupby(
            unknown_records["event_id"]
        ).any().sum()
    )


# Analyze both modules

check_object_consistency(
    module1_input,
    "MODULE 1"
)

check_object_consistency(
    module2_input,
    "MODULE 2"
)

print("\n" + "=" * 65)
print("OBJECT TYPE CONSISTENCY ANALYSIS COMPLETED")
print("=" * 65)
```
