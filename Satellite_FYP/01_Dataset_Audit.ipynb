```python
# ============================================================
# CELL 4: DEFINE PROJECT DIRECTORIES
# ============================================================

# Main project directory
PROJECT_DIR = Path(
    "/content/drive/MyDrive/Smart_Satellite_FYP"
)

# Original dataset directory
RAW_DATA_DIR = PROJECT_DIR / "Raw_Data"

# Extracted dataset directory
EXTRACTED_DIR = RAW_DATA_DIR / "Extracted"

# Audit reports directory
AUDIT_DIR = PROJECT_DIR / "Results" / "Dataset_Audit"

# Create required directories if they do not exist
EXTRACTED_DIR.mkdir(parents=True, exist_ok=True)

AUDIT_DIR.mkdir(parents=True, exist_ok=True)

print("Project directories are ready!")

print("\nProject Directory:", PROJECT_DIR)
print("Raw Data Directory:", RAW_DATA_DIR)
print("Extracted Directory:", EXTRACTED_DIR)
print("Audit Directory:", AUDIT_DIR)
```

```python
# ============================================================
# CELL 5: LOCATE ORIGINAL DATASET ZIP
# ============================================================

zip_path = RAW_DATA_DIR / "Collision Avoidance Challenge - Dataset.zip"

# Check whether the ZIP file exists
if zip_path.exists():

    print("Dataset ZIP file found successfully!")

    print("\nFile Name:", zip_path.name)

    print(
        "File Size:",
        round(zip_path.stat().st_size / (1024 * 1024), 2),
        "MB"
    )

else:

    raise FileNotFoundError(
        "Dataset ZIP file not found! "
        "Please upload it to the Raw_Data folder in Google Drive."
    )
```

```python
# ============================================================
# CELL 6: INSPECT DATASET ZIP CONTENTS
# ============================================================

with zipfile.ZipFile(zip_path, 'r') as zip_ref:

    # Get all files available inside ZIP
    zip_files = zip_ref.namelist()

    print("Total files inside ZIP:", len(zip_files))

    print("\nAvailable Files:\n")

    for file in zip_files:

        print(file)
```

```python
# ============================================================
# CELL 7: EXTRACT DATASET
# ============================================================

with zipfile.ZipFile(zip_path, 'r') as zip_ref:

    # Extract original dataset into project directory
    zip_ref.extractall(EXTRACTED_DIR)

print("Dataset extracted successfully!")

print("\nExtracted location:")
print(EXTRACTED_DIR)
```

```python
from pathlib import Path

# Search all CSV files inside extracted dataset
csv_files = list(EXTRACTED_DIR.rglob("*.csv"))

print("CSV Files Found:")
print("-" * 50)

for i, file in enumerate(csv_files, start=1):
    print(f"{i}. Name: {file.name}")
    print(f"   Path: {file}")
    print()

# Find training dataset
train_files = [
    file for file in csv_files
    if file.name.lower() == "train_data.csv"
]

if len(train_files) == 1:
    TRAIN_PATH = train_files[0]
    print("Training dataset found successfully!")
    print("Actual Path:", TRAIN_PATH)

elif len(train_files) > 1:
    print("Multiple train_data.csv files found.")
    print("Please check their paths.")

else:
    print("train_data.csv was not found.")
    print("Check the CSV filenames displayed above.")
```

```python
import zipfile
from pathlib import Path

# Locate the nested training ZIP
TRAIN_ZIP_PATH = (
    EXTRACTED_DIR
    / "Collision Avoidance Challenge - Dataset"
    / "kelvins_competition_data"
    / "train_data.zip"
)

# Create a separate folder for training data
TRAIN_EXTRACT_DIR = EXTRACTED_DIR / "Training_Data"

TRAIN_EXTRACT_DIR.mkdir(parents=True, exist_ok=True)

# Verify and extract training ZIP
if TRAIN_ZIP_PATH.exists():

    print("Training ZIP found!")
    print("Extracting training dataset...")

    with zipfile.ZipFile(TRAIN_ZIP_PATH, 'r') as zip_ref:
        zip_ref.extractall(TRAIN_EXTRACT_DIR)

    print("Training dataset extracted successfully!")

else:
    raise FileNotFoundError(
        f"Training ZIP not found: {TRAIN_ZIP_PATH}"
    )

# Find training CSV
train_files = list(TRAIN_EXTRACT_DIR.rglob("train_data.csv"))

if len(train_files) == 1:

    TRAIN_PATH = train_files[0]

    print("\nTRAINING CSV FOUND!")
    print("Actual Path:", TRAIN_PATH)

elif len(train_files) > 1:

    raise ValueError("Multiple train_data.csv files found.")

else:

    raise FileNotFoundError(
        "train_data.csv not found inside training ZIP."
    )
```
